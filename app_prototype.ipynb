```python
import requests
from bs4 import BeautifulSoup

# The exact website address with "quotes." included!
url = "https://quotes.toscrape.com/"

response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Look for the quote block
element = soup.find('span', class_='text')

if element:
    print("🎉 SUCCESS! The first quote found on the page is:")
    print(element.text)
else:
    print("Could not find the quote. Double check the URL styling!")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"

# This header tricks the website into thinking we are a normal Chrome browser
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

# We pass the headers inside our request
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

all_quotes = soup.find_all('span', class_='text')

print(f"📊 Status Code: {response.status_code}")
print(f"📊 Found {len(all_quotes)} quotes on this page!\n")

for index, quote in enumerate(all_quotes, start=1):
    print(f"Quote #{index}: {quote.text}")
    print("-" * 40)
```
